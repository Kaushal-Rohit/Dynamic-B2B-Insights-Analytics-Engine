```python
import os

import pandas as pd
import json


def standardize_dataframe(
    df: pd.DataFrame,
    text_column,
    metadata_columns: list = None,
    title_column: str = None,
    save_filename: str = "standardized_data.csv"
) -> pd.DataFrame:
    
    if metadata_columns is None:
        metadata_columns = []

    standardized_rows = []
    
    for _, row in df.iterrows():
        title = ""
        if title_column and title_column in df.columns:
            title = str(row[title_column])
    
        content = ""
        if isinstance(text_column, list):
            content = " ".join([str(row[col]) for col in text_column if col in df.columns])
        elif text_column in df.columns:
            content = str(row[text_column])
            
        metadata = {}

        for col in metadata_columns:
            if col in df.columns:
                value = row[col]
                
                if isinstance(value, pd.Timestamp):
                    value = value.isoformat()

                metadata[col] = value
        
        metadata_json = json.dumps(metadata, default=str)

        standardized_rows.append({
            "title": title,
            "content": content,
            "metadata": metadata_json
        })

    # Create final dataframe
    standardized_df = pd.DataFrame(standardized_rows)

    data_folder = "data"

    if not os.path.exists(data_folder):
        os.makedirs(data_folder)
        print(f"[INFO] '{data_folder}' folder created.")

    save_path = os.path.join(data_folder, save_filename)

    standardized_df.to_csv(save_path, index=False)

    print(f"[INFO] Cleaned data saved at: {save_path}")

    return standardized_df
```

```python
df = pd.read_csv("data.csv ")



clean_data = standardize_dataframe(
    df=df,
    text_column=["Ticket Description", "Ticket Subject", "Ticket Type"],
    title_column="Product Purchased",
    metadata_columns=["Customer Age" , "Customer Gender" , "Date of Purchase" , "Ticket Channel" , "Customer Satisfaction Rating"]
    )

clean_data.head()
```
